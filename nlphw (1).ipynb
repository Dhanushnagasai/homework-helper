```python
import wikipedia

def get_answer(question):
    try:
        return wikipedia.summary(question, sentences=2)
    except wikipedia.exceptions.DisambiguationError as e:
        return f"That topic is too broad. Try one of these: {', '.join(e.options[:5])}."
    except wikipedia.exceptions.PageError:
        return "Sorry, I couldn’t find anything."
    except Exception as e:
        return f"Error: {str(e)}"

def write_essay(topic):
    try:
        content = wikipedia.page(topic).content
        return content[:1500] + "..." if len(content) > 1500 else content
    except:
        return "Couldn't write an essay on this topic."

def analyze_topic(topic):
    try:
        content = wikipedia.summary(topic, sentences=5)
        lines = content.split('. ')
        analysis = "\n".join(f"- {line.strip()}" for line in lines if line.strip())
        return analysis
    except:
        return "Analysis not available for this topic."
```

```python
!pip install duckduckgo-search transformers sympy

from duckduckgo_search import DDGS
from transformers import pipeline
from sympy import sympify

# Initialize transformers pipeline
generator = pipeline("text-generation", model="gpt2", tokenizer="gpt2")

# 🔍 Get short web answer using DuckDuckGo
def get_answer(query):
    with DDGS() as ddgs:
        results = ddgs.text(query)
        for r in results:
            return r.get("body", "No answer found.")
    return "Sorry, I couldn't find an answer."

# 📝 Essay writer
def write_essay(topic):
    prompt = f"Write an essay about {topic}. "
    essay = generator(prompt, max_length=200, num_return_sequences=1)[0]['generated_text']
    return essay.strip()

# 📊 Topic analyzer
def analyze_topic(topic):
    prompt = f"Analyze the topic: {topic}. Provide insights and key points. "
    analysis = generator(prompt, max_length=200, num_return_sequences=1)[0]['generated_text']
    return analysis.strip()

# ➗ Math solver (you can add this if you want math solving too)
def solve_math(expression):
    try:
        result = sympify(expression)
        return f"Solution: {result}"
    except:
        return "❌ Invalid math expression."

# 🤖 Main Chatbot Function
def homework_helper():
    print("📚 Homework Helper Bot\n")
    print("1️⃣ Ask a question")
    print("2️⃣ Write an essay on a topic")
    print("3️⃣ Analyze a topic")
    print("4️⃣ Solve a math problem")
    print("5️⃣ Exit")

    while True:
        choice = input("\nChoose an option (1/2/3/4/5): ")

        if choice == "1":
            q = input("🧑 Your question: ")
            print("🤖", get_answer(q))

        elif choice == "2":
            topic = input("🧑 Topic for essay: ")
            print("\n📝 Essay:\n")
            print(write_essay(topic))

        elif choice == "3":
            topic = input("🧑 Topic to analyze: ")
            print("\n📊 Analysis:\n")
            print(analyze_topic(topic))

        elif choice == "4":
            expression = input("🔢 Enter math expression: ")
            print(solve_math(expression))

        elif choice == "5":
            print("👋 Exiting Homework Helper!")
            break

        else:
            print("❌ Please choose a valid option (1/2/3/4/5).")

# Run the chatbot
homework_helper()
```
